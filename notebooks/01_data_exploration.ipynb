## Olist orders / payments / customers

```python
orders = pd.read_csv(RAW_DIR / 'olist_orders_sample.csv')
payments = pd.read_csv(RAW_DIR / 'olist_order_payments_sample.csv')
customers = pd.read_csv(RAW_DIR / 'olist_customers_sample.csv')

for name, df in [('orders', orders), ('payments', payments), ('customers', customers)]:
    print(name, df.shape)
    display(df.head())
```

```python
orders['order_status'].value_counts()
```

```python
payments['payment_type'].value_counts()
```

```python
customers['customer_state'].value_counts()
```

## Quick joinability check

Confirm every payment has a matching order, and every order has a matching customer — the same checks `src/data_cleaning.py` performs.

```python
orphan_payments = set(payments['order_id']) - set(orders['order_id'])
orphan_customers = set(orders['customer_id']) - set(customers['customer_id'])
print('orphan payments:', len(orphan_payments))
print('orders with missing customer:', len(orphan_customers))
```
